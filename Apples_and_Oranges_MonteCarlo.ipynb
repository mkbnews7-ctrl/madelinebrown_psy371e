### Monte Carlo Simulation of Sample Means

This simulation demonstrates the Central Limit Theorem by repeatedly taking samples from a population and plotting the distribution of their means. Even if the original population distribution is not normal, the distribution of sample means tends to be normal for a sufficiently large sample size.

```python
import matplotlib.pyplot as plt
import numpy as np

# Generate some synthetic data for demonstration (e.g., apple weights)
# Let's assume apple weights are normally distributed with a mean of 150g and std dev of 10g.
np.random.seed(42) # for reproducibility
apple_data = np.random.normal(loc=150, scale=10, size=10000)

# Example: simulating 2000 sample means from the dataset
n_simulations = 2000
sample_size = 30

apple_means = []
# orange_means = []

for _ in range(n_simulations):
  # Take a random sample from your apple data and find its mean
  sample = np.random.choice(apple_data, size=sample_size)
  apple_means.append(np.mean(sample))

# Plotting the distribution
plt.figure(figsize=(10, 6))
plt.hist(apple_means, bins=30, alpha=0.7, label='Apples (Sample Means)', color='skyblue', edgecolor='black')
plt.title('Monte Carlo Simulation of Sample Means of Apple Weights')
plt.xlabel('Mean Weight (grams)')
plt.ylabel('Frequency')
plt.legend()
plt.grid(axis='y', alpha=0.75)
plt.show()
```
