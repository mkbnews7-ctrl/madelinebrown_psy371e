## Object-Oriented Programming (OOP) with a `Dog` Class

Object-Oriented Programming (OOP) is a programming paradigm based on the concept of "objects", which can contain data and code. In Python, you define objects using **classes**.

A **class** is a blueprint for creating objects (instances). It defines a set of attributes (data) and methods (functions) that the created objects will have.

Let's create a `Dog` class with attributes like `name` and `breed`, and methods like `speak`, `sit`, `bark`, and `stand`.

```python
class Dog:
    def __init__(self, name, breed):
        """Initializes a new Dog object."""
        self.name = name
        self.breed = breed
        print(f"A new dog named {self.name} ({self.breed}) has been created!")

    def speak(self, sound="Woof!"):
        """Makes the dog speak a given sound."""
        return f"{self.name} says {sound}"

    def sit(self):
        """Makes the dog sit."""
        return f"{self.name} is now sitting."

    def bark(self):
        """Makes the dog bark."""
        return f"{self.name} barks loudly! Arf! Arf!"

    def stand(self):
        """Makes the dog stand up."""
        return f"{self.name} stands up on all fours."
```

Now that we have defined our `Dog` class, we can create instances (objects) of it. Each instance will have its own `name` and `breed`, and can perform the actions defined by its methods.

```python
piper = Dog("Piper", "Chihuahua")
print(piper.speak())
print(piper.sit())
print(piper.bark())
print(piper.stand())
```

```python
# Create a Dog object
my_dog = Dog("Buddy", "Golden Retriever")

# Call methods on the dog object
print(my_dog.speak())
print(my_dog.bark())
print(my_dog.sit())
print(my_dog.stand())

# Create another Dog object
another_dog = Dog("Lucy", "Beagle")

# Lucy can also do dog things!
print(another_dog.speak("Yap!"))
print(another_dog.sit())
```
